# Set Comprehesion 

```python
#set comprhesion:
numbers=[1,2,3,4,5]
squares={x*x for x in numbers}
print(squares)
```

```python
# Find the even number from same list :
numbers=[1,2,3,4,5]
even={i for i in numbers if i%2==0}
print(even)
```

```python
#square 1 to 5 ;
numbers=[1,2,3,4,5]
square=set()
for i in numbers:
        square.add(i*i)
print(square)        
    
```
