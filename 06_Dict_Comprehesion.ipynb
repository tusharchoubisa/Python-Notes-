# Dictionary Comprehesion 

```python
numbers=[1,2,3,4,5,6]
squars={x:x*x  for x in numbers }
print(squars)
```

```python
# Find the squar
numbers=[1,2,3,4,5,6]
squars={x:x*x  for x in numbers }
print(squars)
```

```python
#Find the Cube
numbers=[1,2,3,4,5,6]
cube={x:x*x*x  for x in numbers }
print(cube)
```

```python
# Print even/odd after keys 
numbers=[1,2,3,4,5,6]
result={x:'even' if x%2==0 else'odd' for x in numbers}
print(result)
```

```python
#print even after kehys 
numbers=[1,2,3,4,5,6]
result={x:x*x for x in numbers if x%2==0}
print(result)
```
