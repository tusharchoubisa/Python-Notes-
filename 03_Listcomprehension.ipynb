```python
# normal Code :
marks=[20,40,60,50,65,80,90]
print(marks)
```

```python
#normal code 
marks=[20,40,60,50,65,80,90]
new_marks = []
for x in marks:
    new_marks.append(x+2)
print(new_marks)    
```

```python
# List comprehensions 
marks=[20,40,60,50,65,80,90]
new_marks = [x+2 for x in marks]
print(new_marks)
```

```python
# find the Cubes 
# Normal code 
cubes=[]
for x in range(10):
    if x% 2==0:
        cubes.append(x**3)
print(cubes)
```

```python
# Cube using Comprehension 
cubes=[x**3 for x in range(10) if x%2==0]
print(cubes)
```

```python
#1 to 5 Square
num=[1,2,3,4,5]
square=[]
for i in num:
    square.append(i*i)
print(square)
```
