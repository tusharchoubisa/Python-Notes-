# Set :
It is a collection of Unique items.  
It is used to eliminate duplicate items from a list.    
ALso, it supports set operation like union itersection and difference.      
Seta are mutable and no garanty will return output in ordering or order format , remove dduplicate value automatically.

# Creating Set 
In Python sets are written with curly brackets.

```python
age={11,12,13,14,31,24,23,21,12}
print(age)
```

```python
set=set([1,2,3,4,5,6,7,8])
print(set)
```

```python
# Create a set using different data types
set={12,12,'tushar','Python',90,85,95,'harsh'}
print(set)
```

# Set Method   
change items    
Once a set is created you cannot modify items, but you can add new items.

# 1.Add() items    
Use add() method to add one items to a set.

```python
set={'Pyhton','SQL'}
set.add('devops')
print(set)
```

# 2.Update() 
Use Update() method to add more than one items.

```python
set={'Pyhton','SQL'}
set.update(['devops','Cloud Computing'])
set
```

# 3.Length():

```python
print(len(set))
```

# 4.Remove()
use to remove() method to remove an items in a set 

```python
set={'Cloud Computing', 'Pyhton', 'SQL', 'devops'}
set.remove('SQL')
print(set)
```

```python
# Removeing the miltiple value without using remove() value.
set={'AWS','Python','DBMS','Machine learning','NLP','AI','Cloud Computing','NLP','EDA','Linix','CI-CD','krubernte'}
remove={'Machine learning','NLP','AI','NLP','EDA'}
for i in remove: 
    set.remove(i)
print(set)    
```

```python
set={'AWS','Python','DBMS','Machine learning','NLP','AI','Cloud Computing','NLP','EDA','Linix','CI-CD','krubernte'}
set.discard('NLP')
print(set)
```

# 5.Pop()

```python
set={'AWS','Python','DBMS','Machine learning','NLP','AI','Cloud Computing','NLP','EDA','Linix','CI-CD','krubernte'}
set.pop()
```
