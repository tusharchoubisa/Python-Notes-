# Nested Loop 

```python
for i in range(5):
    for j in range(5):
        print('*',end=' ')
    print()    
```

```python
new_list=[i for i in range(11) if i%2==0]
print(new_list)
```

```python
# Make a list 0 number commaon to both the list using list Comprehesion :
list_1=[102,132,143,243,256,276]
list_2=[102,154,567,876,980,435]
common_num=[a for a in list_1 for b in list_2 if a==b]
print(common_num)
```

```python
list_1=[102,132,143,243,256,276]
list_2=[102,154,567,876,980,435]
common_list=[]
for a in list_1:
    for b in list_2:
        if a==b:
            common_list.append(a)
print(common_list)                 
```
